```python

# 建立 Gemini 模型
model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",  # 使用 Gemini 2.5 Flash 模型
    temperature=0.7,           # 控制回應的創造性 (0-1)
    top_p=0.9,                # 控制回應的多樣性
    max_output_tokens=1000,   # 限制回應長度
)

print("✅ Gemini 模型建立成功！")
print(f"📋 使用模型：gemini-2.5-flash")
print(f"🌡️ 溫度設定：{model.temperature}")
print(f"🎯 Top-p 設定：{model.top_p}")
print(f"📏 最大輸出長度：{model.max_output_tokens} tokens")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda

```

```python
prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一個專業的客服代表，負責回覆客戶郵件。請用友善、專業的語調回覆。"),
    ("human","請回覆這封客戶郵件：\n\n{email_content}")
])
```

```python
def format_email_reply(ai_answer:str) -> str:
    import datetime
    date_str = datetime.datetime.now().strftime("中華民國,西元%Y-%m-%d %H:%M")
    formatted_reply = f"""{ai_answer}

感謝您的來信，如有其他問題請隨時聯繫我們。

此致
致理客服團隊

回覆時間:{date_str}
"""

    return formatted_reply

format_email = RunnableLambda(format_email_reply)
```

```python
chain = (prompt_template 
        | model
        | StrOutputParser()
        | format_email
)
```

```python
customer_email = """您好:

我最近購買了貴公司的產品，但是發現包裝有損壞，而且產品功能與說明書不符。
請問可以退貨或換貨嗎？我已經購買一週了。

期待您的回覆。


王小明"""
```

```python
result = chain.invoke({
    "email_content":customer_email
})

print(result)
```
